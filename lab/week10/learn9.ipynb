### GUI 프로그래밍

* 윈도우, 메뉴, 버튼, 마우스를 사용하여 사용자와 정보를 교환하는 사용자 인터페이스

### tkinter
* 파이썬에서 그래픽 사용자 인터페이스를 개발할 때 필요한 모듈

#### tkinter 시작하기
* from tkinter import* <br>
-> tkinter 모듈을 포함시키는 것이다.
* window = Tk()<br>
-> Tk()을 호출하면 Tk 클래스의 객체가 생성되면서 화면에 하나의 윈도우가 생성된다.
* label = Label(window,text="Hello tkinter")<br>
-> 레이블 위젯을 생성한다.
* label.pack()<br>
-> pack()은 압축 배치 관리자를 이용하여서 레이블을 컨테이너에 위치시킨다.
* window.mainloop()
-> 루트 윈도우의 mainloop()는 이벤트 처리 루프로서 사용자로부터 오는 마우스나 이벤트를 처리한다.

```python
from tkinter import*

window = Tk()
label=Label(window,text="Hello tkinter")
label.pack()

window.mainloop()
```

#### 위젯들
* button : 간단한 버튼으로 명령을 수행할 때 사용된다. <br>
* Canvas : 화면에 무언가를 그릴 때 사용한다. <br>
* Checkbutton : 2가지 구별되는 값을 가지는 변수를 표현한다. <br>
* Entry : 한 줄의 텍스트를 입력받는 필드이다. <br>
* Frame : 컨테이너 클래스이다. 프레임은 경계선과 배경을 가지고 있다. 다른 위젯들을 그룹핑하는데 사용된다. <br>
* Label : 텍스트나 이미지를 표시한다. <br>
* Listbox : 선택 사항을 표시한다. <br>
* Menu : 메뉴를 표시한다. 풀다운 메뉴나 팝업 메뉴가 가능하다.<br>
* Menubutton : 메뉴 버튼이다. 풀다운 메뉴가 가능하다. <br>
* Message : 텍스트를 표시한다. 레이블 위젯과 비슷하다. 하지만 자동적으로 주어진 크기로 텍스트를 축소할 수 있다.<br>
* Radiobutton : 여러 값을 가질 수 있는 변수를 표시한다. <br>
* Scale : 슬라이더를 끌어서 수치를 입력하는데 사용된다. <br>
* scrollbar : 캔버스, 엔트리, 리스트 박스, 텍스트 위젯을 위한 스크롤 바를 제공한다. <br>
* Text : 형식을 가지는 텍스트를 표시한다. 여러가지 스타일과 속성으로 텍스트를 표시할 수 있다. <br>
* Toplevel : 최상위 윈도우로 표시되는 독립적인 컨테이너 위젯이다. <br>
* LacelFrame : 경계선과 제목을 가지는 프레임 위젯의 변형이다. <br>
* PanedWindow : 자식 위젯들을 크기조절이 가능한 패널로 관리하는 컨테이너 위젯이다. <br>
* Spinbox : 특정한 범위에서 값을 선택하는 엔트리 위젯의 변형 <br>

#### 컨테이너 종류
* Fram 프레임e () tk.Frame
  * 가장 기본적이고 널리 쓰이는 투명한 컨테이너.
* LabelFrame(라벨 프레임) tk.LabelFrame
  * Frame에 '테두리 선'과 '제목(라벨)'이 추가된 형태.
* Toplevel(탑레벨) tk.Toplevel
  * 메인 윈도우와 별개로 완전히 독립된 새로운 보조창(팝업창)을 띄우는 컨테이너.
* PaneWindow(페인드 윈도우)tk.PanedWindow
  * 내부에 배치된 위젯들 사이에 마우스로 드래그하여 크기를 조절할 수 있는 구분선(경계선)을 제공 
* Notebook(노트북/탭) ttk.Notebook
  * 웹 브라우저의 탭이나 책의 색인처럼 여러개의 페이지(Tab)를 전환하며 볼수 있게 해주는 컨테이너(일반 tk가 아닌 tkinter.ttk 모듈에 포함)
* Canvas(캔버스) tk.Cansvas
  * 선, 원, 도형을 그리거나 이미지를 배치하는 도화지 위젯이지만, 일반 위젯을 내부에 얹을 수 있는 컨테이너 역할도 겸함.

 #### 배치 관리자
 * 압축(pack) 배치 관리자 
    * 위젯(버튼, 레이블 등)을 위젯을 화면에 배치할 때 구체적인 좌표를 지정하지 않고, 마치 물건을 상자 안에 차곡차곡 압축하여 밀어넣듯이 배치하는 방식
 * 격자(grid) 배치 관리자
    * 위젯을 테이블 형태로 배치
 * 절대(place) 배치 관리자
    * 위젯을 테이블 형태로 배치

```python
#온도 변환기
from tkinter import*

#이벤트 처리 함수를 정의한다.
def process():
    tf = float(e1.get()) #e1에서 문자열을 읽어서 부동소수점형으로 변경
    tc = (tf-32.0)*5.0/9.0 #화씨를 섭씨로 변환
    e2.delete(0, END) #처음부터 끝까지 지운다.
    e2.insert(0, str(tc)) #tc 변수의 값을 문자열로 변환하여 추가

window = Tk()

Label(window, text="화씨").grid(row=0, column=0)
Label(window, text="섭씨").grid(row=1, column=0)

e1 = Entry(window)
e2 = Entry(window)
e1.grid(row=0, column=1)
e2.grid(row=1, column=1)

Button(window, text="화씨 -> 섭씨", command=process).grid(row=2, column=1)
window.mainloop()
```

```python
#숫자 추측 게임
from tkinter import*
import random

answer = random.randint(1, 100) #1~100 사이의 난수 발생

def guessing():
    guess = int(guessField.get()) #텍스트 필드에서 사용자가 입력한 값을 가져옴

    if guess < answer:
        msg = "높음!"
    elif guess > answer:
        msg = "낮음!"
    else:
        msg = "정답!"

    resultLabel["text"]=msg #메시지 출력
    guessField.delete(0, 5)
def reset(): #정답을 다시 설정
    global answer
    answer = random.randint(1, 100)
    resultLabel["text"]="다시 한번 하세요!"

window = Tk()
window.configure(bg="white")
window.title("숫자를 맞춰보세요!")

window.geometry("500x80")
titleLabel=Label(window, text="숫자 게임에 오신 것을 환영합니다!",bg="white")
titleLabel.pack()

guessField = Entry(window)
guessField.pack(side="left")
tryButton = Button(window, text="시도", fg="green",bg="white",command=guessing)
tryButton.pack(side="left")

resetButton = Button(window, text="초기화", fg="red",bg="white",command=reset)
resetButton.pack(side="left")
resultLabel = Label(window, text="1부터 100사이의 숫자를 입력하시오.",bg="white")
resultLabel.pack(side="left")

window.mainloop()
```

```python
#가위 바위 보 게임
import random
from tkinter import*

window = Tk()
Label(window, text="선택하세요",font=("Helevetica", 16)).pack()
frame = Frame(window)

#이미지를 d드라이브로 복사하여 주세요!
rock_image = PhotoImage(file=r"C:\\Users\\qls25\\OneDrive\\사진\\카메라 앨범\\다운로드.png").subsample(5,5)
paper_image = PhotoImage(file=r"C:\\Users\\qls25\\OneDrive\\사진\\카메라 앨범\\다운로드 (2).png").subsample(5,5)
scissors_image = PhotoImage(file=r"C:\\Users\\qls25\\OneDrive\\사진\\카메라 앨범\\다운로드 (1).png").subsample(5,5)

def pass_s():
    decide("가위")
def pass_r():
    decide("바위")
def pass_p():
    decide("보")

rock = Button(frame, image=rock_image, command=pass_r)
rock.pack(side="left")
paper = Button(frame, image=paper_image, command=pass_p)
paper.pack(side="left")
scissors = Button(frame, image=scissors_image, command=pass_s)
scissors.pack(side="left")

frame.pack()
Label(window, text="컴퓨터는 다음을 선택하였습니다.",font=("Helevetica", 16)).pack()

computer_image = Label(window, image=rock_image)
computer_image.pack()
output = Label(window, text="",font=("Helvetica",16))
output.pack()

def decide(human):
    computer = random.choice(["가위","바위","보"])
    if computer == "바위":
        computer_image["image"]=rock_image
    elif computer == "보":
        computer_image["image"]=paper_image
    else:
        computer_image["image"]=scissors_image

    if (computer == "바위" and human == "보") or (computer == "보" and human == "가위") or (computer == "가위"and human == "바위"):
        result = "인간 승리!"
    elif computer == human:
        result = "무승부!"
    else: 
        result = "컴퓨터 승리!"
    output.config(text="인간: "+human+" 컴퓨터: "+computer+" "+result)

window.mainloop()
```

```python
# TIC-TAC-TOE 게임

from tkinter import*

#i번째 버튼을 누를 수 있는지 검사한다. 누를 수 있으면 X나 O를 표시함.
def checked(i):
    global player
    button = list[i]    #리스트에서 i번째 버튼 객체를 가져옴.

    # 버튼이 초기상태가 아니면 이미 누른 버튼이므로 아무것도 하지 않고 리턴함.

    if button["text"] != "   ":
        return
    button["text"] = "  "+player+"   "
    button["bg"] = "yellow"
    if player == "X":
        player = "O"
        button["bg"]="yellow"
    else:
        player="X"
        button["bg"] = "lightgreen"

window = Tk()   #윈도우를 생성한다.
player="X"      #시작은 플레이어 X이다.
list=[]

#9개의 버튼을 생성하여 격자 형태로 윈도우에 배치함.
for i in range(9):
    b=Button(window, text="   ",command=lambda k=i: checked(k))
    b.grid(row=i//3, column=i%3)
    list.append(b) #버튼 객체를 리스트에 저장함.

window.mainloop()
```

```python
# 그림판 프로그램 만들기
from tkinter import*
from tkinter.colorchooser import askcolor

window = Tk()
window.title("그림판 프로그램")

old_x, old_y = None,None
mode = "pen"
mycolor = "red"

def use_pen():
    global mode
    mode = "pen"

def use_brush():
    global mode
    mode = "brush"

def choose_color():
    global mycolor
    color = askcolor()
    if color[1]:
        mycolor = color[1]

def use_eraser():
    global mode
    mode = "erase"

def clear_all():
    canvas.delete(ALL)

#각 마우스 이벤트가 발생하면 다음과 같은 함수들이 호출됨.
def paint(event):
    global var, eraes_on, old_x, old_y
    fill_color = "white"if mode == "erase" else mycolor
    if old_x and old_y:
        canvas.create_line(old_x, old_y, event.x,event.y,capstyle=ROUND, width=var.get(),fill=fill_color)

    old_x = event.x
    old_y = event.y

def reset(event):
    global old_x,old_y
    old_x, old_y = None,None


penButton = Button(window,text='펜',command=use_pen)
penButton.grid(row=0, column=0, sticky=W+E)

brushButton = Button(window,text='브러쉬',command=use_brush)
brushButton.grid(row=0, column=1, sticky=W+E)

colorButton = Button(window,text='색상선택',command=choose_color)
colorButton.grid(row=0, column=2, sticky=W+E)

eraseButton = Button(window,text='지우개',command=use_eraser)
eraseButton.grid(row=0, column=3, sticky=W+E)

clearButton = Button(window,text='모두 삭제',command=clear_all)
clearButton.grid(row=0, column=4, sticky=W+E)

var = DoubleVar()
scale = Scale(window,variable=var,orient=VERTICAL)
scale.grid(row=1,column=5,sticky=N+S)

canvas = Canvas(window, width=600, height=400, bg='white')
canvas.grid(row=1, column=0, columnspan=5)

canvas.bind('<B1-Motion>',paint)
canvas.bind('<ButtonRelease-1>',reset)

window.mainloop()
```

```python
# 공 애니메이션 만들기 -1

from tkinter import*
import time
import random

window = Tk()

canvas = Canvas(window,width=600,height=400)
canvas.pack()

class Ball():
    def __init__(self,color,size):
        self.id=canvas.create_oval(0,0,size, size,fill=color)
        self.dx=random.randint(1,10)
        self.dy=random.randint(1,10)

    def move(self):
        canvas.move(self.id,self.dx,self.dy)
        x0, y0, x1, y1=canvas.coords(self.id)
        if y1 > canvas.winfo_height()or y0 < 0: #원이 위쪽이나 아래쪽을 벗어났으면
            self.dy = -self.dy          #dy의 부호를 반전시킨다.

        if x1 > canvas.winfo_width() or x0 < 0: #원이 왼쪽이나 오른쪽으로 벗어났으면
            self.dx = -self.dx

ball1=Ball("blue",60)
ball2=Ball("green",100)
ball3=Ball("orange",80)

def animate():
    ball1.move()
    ball2.move()
    ball3.move()
    window.update()
    window.after(50,animate)

window.after(50,animate)
window.mainloop()

        
```

```python
# 공 애미메이션 -2

from tkinter import*
import time
import random

window = Tk()
canvas = Canvas(window, width=600, height=400)
canvas.pack()

class Ball():
    def __init__(self,color,size):
        self.id=canvas.create_oval(0,0,size,size,fill=color)
        self.dx=random.randint(1,10)
        self.dy=random.randint(1,10)

    def move(self):
        canvas.move(self.id,self.dx,self.dy)
        x0,y0,x1,y1=canvas.coords(self.id)
        if y1 > canvas.winfo_height() or y0 < 0: #원이 위쪽이나 아래쪽으로 벗어났으면
            self.dy = -self.dy       #dy의 부호를 반전시킨다.
        if x1 > canvas.winfo_width() or x0 < 0:  #원이 왼쪽이나 오른쪽으로 벗어났으면
            self.dx = -self.dx       #dx의 부호를 반전시킨다.

colors = ["red","orange","yellow","green","blue","indigo","violet"]
ballList = []
for i in range(30):
    ballList.append(Ball(random.choice(colors),60))

def animate():
    for i in range(30):
        ballList[i].move()
    window.after(50, animate)

window.after(50, animate)
window.mainloop()
```
