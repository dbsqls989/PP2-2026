# counter 클래스 만들기
- 데이터 : 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기


```python
class Counter :
    def __init__(self, initialcount=0): #생성함수
        self.count = initialcount 

    def increment(self, amount= 1): 
        self.count += amount

    def get_count(self):
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())

    myc.increment()
    print(myc.get_count())

    myc.increment()
    print(myc.get_count()) #myc.count=2

    for _ in range(10):
        myc.increment() #원래 myc.count=2, 2부터 반복 myc.count+=1 10번반복
    print(myc.get_count()) #myc.count=12

    yourc=Counter(5)

    print(yourc.get_count())
    #yourc.count=5 출력
    yourc.increment() #5+=1
    print(yourc.get_count()) #6으로 출력
    yourc.increment(4) #10
    print(yourc.get_count()) #10으로 출력


if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터: 현재 채널 번호, 볼륨량, on여부
- 동작: 켜기, 끄기, 채널변경, 햔재 TV 객체 내용 출력

- 클래스명: MyTelevision

```python
class MyTelevision:
    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on

    def turn_on(self):
            self.is_on = True # 전원 켜기

    def turn_off(self):
            self.is_on = False # 전원 끄기

    def change_channel(self, channel):
            self.channel = channel

    def change_volume(self, volume):
            self.volume = volume
            
    def display_status(self):
            print(f"Channel: {self.channel}, Volume: {self.volume}, Is on: {self.is_on}")

def test_television():
    mytv = MyTelevision(channel=11)
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()

if __name__ == "__main__":
    test_television()

```

# Tv 리모콘 클래스
- 데이터는 없고 동작만 있음
- 연관된 TV를 키거나 끄라는 동작만 지원
- TV 상태 표시 추가

```python
class TVRemocon:
    def __init__(self, tv:MyTelevision):
        self.tv = tv

    def press_on(self):
        self.tv.turn_on()

    def press_off(self):
        self.tv.turn_off()

    def press_display(self):
        self.tv.display_status()

def test_remocon():
    tv1 = MyTelevision()
    rc1 = TVRemocon(tv1)
    rc1.press_on() #true로 상태 변경
    rc1.press_display()
    rc1.press_off()
    rc1.press_display()

if __name__ == "__main__":
    test_remocon()
```
