### 파일의 논리적인 구조

* 파일: 바이트들이 순차적으로 저장되어 있고 맨 끝에는 EOF(end-od-file) 마커가 있다.<br>
* 파일 포인터: 각 바이트는 첫 바이트는 0, 두번째 바이트는 1등으로 offset이 부여된다.
* 파일의 종류
* 텍스트 파일: 각 바이트가 문자로 구성된 파일(메모장으로 열었을 떄 눈으로 글자 확인 가능)
    * 한글 유니코드는 각 문자가 2B로 구성되어 있으며 저장될 때는 UTF-8등으로 안코딩되어 저장되고 읽을 때는 디시 디코딩되어 읽힘
 
* 이진 파일: 이진수의 바이트 배열이 저장된 파일(메모장으로 열면 글자 깨져 보임) 


### 디렉토리(폴더)의 개념
* 폴더(디렉토리)
    * 폴더는 파일이나 서브폴더들을 가질 수 있다. <br>
    * 폴더는 일반적으로 디렉토리라고 한다. <br>
    * 루트디렉토리는 '/' 윈도우에서는 '\' 이다. <br>
    * 드라이브(혹은 볼륨)는 윈도우에서 별도의 루트디렉토리를 가진 파일 시스템을 의마한다.(C:, D: 등) <br>
    * 부모폴더와 자식폴더 사이에는 '/' 문자로 구분한다. (C:\user\song\PP2-2026) <br>

### 경로(Path) 개념
* 경로
    * 절대경로: C:\user\song\PP2-2026\chap10\input.txt
    * 현재경로: 현재 프로그램이 실행 중인 절대경로
    * 상대경로: .\chap10\input.txt 처럼 현재경로를 기준으로 한 경로
        * ".": 현재 디렉토리("C:\user\song\PP2-2026"처럼 현재 실행 중인 python 파일 폴더)
        * "..": 부모 디렉토리("..\chap10\input,txt)
