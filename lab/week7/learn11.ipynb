#### abs()함수
* abs() 함수는 숫자의 절대값을 반환하는 데 사용된다.

```python
# abs()함수 예제
i = 20
abs(i)
20

f=-30.92
abs(f)
30.92
```

#### all()함수
* all()함수는 시퀀스(리스트나 딕셔너리 등)를 받아서 시퀀스의 모든 항목이 참이면 True를 반환한다. 그렇지 않으면 False 반환. 0이 아닌 값은 참으로 간주되고 0은 거짓으로 간주

```python
# all()함수 예제
mylist = [1,3,4,6]
all(mylist)
True

mylist = [1,3,4,0]
all(mylist)
False
```

#### any()함수
* any()함수는 시퀀스 객체에 있는 한 개의 항목이라도 참인 경우 참을 반환한다. 그렇지 않으면 거짓 반환

```python
# any()함수 예제
mylist = [0, 1, 2, 3]
any(mylist)
True
```

#### eval()함수
* eval() 함수는 전달된 수식을 구문 분석하고 프로그램 내에서 수식의 값을 반환한다.

```python
# eval함수 예제
x = 10
eval('x+1')
11

exp = input("파이썬 수식을 입력하시오: ")
# 파이썬 수식을 입력하시오: 2**10
eval(exp)
1024
```

#### sum()함수
* sum() 함수는 리스트에 존재하는 항목들을 전부 더하여 합계를 반환한다.

```python
# sum()함수 예제

sum([1,2,3])
6
sum([1,2,3],20)
26
```

#### len()함수
* len()함수는 객체의 길이를 계산하여 반환하는 함수이다.

```python
len("All's well that ends well.")
27
len([1,2,3,4,5])
5
```

#### list()함수
* 리스트를 생성하는 함수

```python
s = 'abcdefg'
list(s)
['a','b','c','d','e','f','g']

t=(1,2,3,4,5,6)
list(t)
[1,2,3,4,5,6]
```

#### map()함수
* map()함수는 반복가능한 객체(리스트,튜플 등)의 각 항목에 주어진 함수를 적용한 후 적용 결과를 반환한다.

```python
# map()함수 예제
def square(n):
    return n*n
mylist=[1,2,3,4,5]
result = list(map(square, mylist))
print(result)
```

#### max(),min() 함수
* max() 함수: 리스트나 튜플, 문자열에서 가장 큰 항목을 반환한다.
* min() 함수: 가장 작은 항목을 반환한다.

```python
# min(),max()함수 예제

values = [1,2,3,4,5]
max(values)
5
min(values)
1
```

#### enumerate()함수
* 시퀀스 객체를 입력 받아, 열거형 (enumerate) 객체를 반환한다.


```python
# enumerate()함수 예제
seasons = ['Spring','Summer','Fall','Winter']

list(enumerate(seasons))
[(0,'Spring'),(1,'Summer'),(2,'Fall'),(3,'Winter')]
list(enumerate(seasons,start=1))
[(1,'Spring'),(2,'Summer'),(3,'Fall'),(4,'Winter')]
filter()
```

#### filter()함수
* filter()함수는 특정 조건을 만족하는 요소만 뽑는다

```python
def myfilter(x):
    return x > 3

result = filter(myfilter,(1,2,3,4,5,6))
print(list(result))
```

#### zip()함수
* zip()함수는 2개의 시퀀스 자료형을 하나로 묶어주는 함수이다.

```python
# zip()함수 예제 -1
numbers = [1,2,3,4]
slist = ['one','two','three','four']
list(zip(numbers,slist))
[(1,'one'),(2,'two'),(3,'three'),(4,'four')]
```

```python
# zip()함수 예제 -2
names = ["KIM","LEE","PARK"]
scores = [100,99,80]

for n, s in zip(names, scores):
    print(n,s)
```

#### 내장함수 예제

```python
invitations = ["Kim","Lee","Park","Choi"]
persons = [1,3,0,6]
sum(persons)
10
# 파티에 한 사람이라도 오는 지 확인 -> any()
any(persons)
True
#모든 초대받은 그룹이 전부 오는 지를 확인 -> all()
all(persons)
False
#가장 많이 오는 그룹에는 몇 사람이나 있는지를 알아보기 -> max()
max(persons)
6
# 이름과 사람수를 묶어서 출력
for name, person in zip(invitations, persons):
    print(name,person)
```

#### 정렬과 탐색
* 파이썬 리스트는 sort()라는 메소드를 가지고 이 메소드는 리스트를 정렬된 상태로 변경한다.
* sorted()라는 내장 함수는 반복 가능한 객체로부터 정렬된 리스트를 새롭게 생성한다.

```python
sorted([4,2,3,5,2,1])
[1,2,3,4,5]

myList = [4,2,3,5,1]
myList.sort()
myList
[1,2,3,4,5]
```

```python
# 예제-2
class Student:
    def __init__(self,name,grade,number):
        self.name = name
        self.grade = grade
        self.number = number
    def __repr__(self):
        return repr((self.name,self.grade,self.number))

students = [
    Student('홍길동',3.9,20160303),
    Student('김철수',3.0,20160302),
    Student('최자영',4.3,20160301),
]
print(sorted(students, key=lambda student: student.number))
```
