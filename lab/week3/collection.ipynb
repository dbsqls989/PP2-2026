## tuple

1. 튜플 (tupple)은 (...)을 사용하여 정의한다.
2. 튜플과 리스트의 차이점은?
- 튜플(tuple)은 변경이 불가능하다.
3. 튜플과 리스트<br>
* 리스트 -> 튜플 생성<br>
ex) myList = [1, 2, 3, 4]<br>
    myTuple = tuple(myList)    #tuple()은 튜플을 생성하는 함수임.<br>
    myTuple <br>
    (1, 2, 3, 4) <br>
* 튜플 -> 리스트 생성<br>
ex) myTuple = (1, 2, 3, 4)<br>
    myList = list(myTuple)     #list()는 리스트를 생성하는 함수이다.<br>
    myList<br>
    [1,2,3,4]<br>
4. 튜플 패킹과 언패킹
* 튜플 패킹: 여러 개의 값을 쉼표(,)로 구분. 소괄호 안에 묶어 하나의 튜플 객체로 생성<br>
    ex) x = ("apple", "banana","grape") <br>
* 튜플 언패킹: 튜플에 담긴 여러 요소를 한 번에 여러 변수에 할당하는 것<br>
    ex) (s1, s2, s3) = x       
5. 튜플의 장점
    * 튜플은 변경 불가능 객체이기 때문에 튜플을 통하여 반복하는 것이 리스트보다 빠르다. 그만큼 처리 코드가 간단해진다.
    * 변경 불가능한 요소들로 이루어진 튜플은 딕셔너리의 키가 될 수 있다.
    * 특정 데이터를 변경하는 것을 금지하려면 튜플을 사용하면 된다.



```python
# 튜플 예제
student = ("홍길동",20,175)
print(student)
```

## set

1. 세트(set)란?
    * 고유한 값들을 저장하는 자료구조 (수학에 집합이라 생각하면 편함).
    * {...}을 사용하여 정의. <br>
    ex) numbers = {1, 2, 3}
2. 세트의 특징
    * 세트는 특정 순서를 유지할 필요가 없으므로 세트 연산은 리스트 연산보다 훨씬 빠르다.
    순서가 없는 항목들의 집합을 원한다면 최선의 선택지가 된다. but 중복된 항목은 없어야함.

```python
# set 예제
fruits = ["apple", "banana", "apple", "orange", "banana"]

fruits_set = set(fruits)

print(fruits_set)
```

## dictionary

1. 딕셔너리(dictionary)란?
    * 키(key)와 값(value)이 1:1로 대응하는 매핑(mapping) 자료형이다.<br>
    ex) 국가 이름(키),수도(값)<br>
2. {...}를 사용하여 정의. 중괄호 안에 키와 값 사이를 쉼표로 분리 시켜야한다.<br>
    ex) dict = {"Name":"홍길동","Age":7,"Class":"초급"}<br>
    
3. 딕셔너리에서 가장 중요한 연산은?
    * 키를 가지고 연관된 값을 찾는 것!<br>
    ex) capitals = {"Korea":"Seoul","USA":"Washington","UK","London"}<br>
        capitals["Korea"]<br>
        Seoul<br>

4. 딕셔너리 함축<br>
    * 중괄호 {..} 안에 키와 값(key: value)을 적고, 뒤에 for 문을 적는다.<br>
    ex) {키: 값 for 변수 in 반복가능객체}

```python
# 딕셔너리 함축 예시
fruits = ["apple","orange","banana"]
dic = {f:len(f)for f in fruits}
print(dic)
```

# byte, bytes, bytearray 공부

- byte: 컴퓨터 데이터의 기본 저장 단위. 8bit => 1byte
- bytes: 불변한 바이트 시퀀스.
- bytearray: 변경 가능한 바이트 시퀀스.

```python
# bytes 예제
data = bytes([65, 66, 67])

print(data)
```

```python
# bytearray 예제
data = bytearray([65, 66, 67])

data[0] = 68

print(data)
```

all(),any() 공부하기

- all() : 모두 참이어야 True 출력
- any() : 하나라도 참이면 True 출력

```python
# all()예제
a = [True, True, True]

print(all(a))
```

```python
# any() 예제
a = [False, True, False]

print(any(a))
```
