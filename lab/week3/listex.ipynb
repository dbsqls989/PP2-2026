## List 연습 노트 파일입니다.

##### 리스트의 합병 연산 (+)<br>리스트의 곱 연산 (*)

```python
fruit1 = ["사과","바나나"]
fruit2 = ["복숭아","포도"]
print (fruit1)
```

##### 리스트 방문<br>
-인덱스를 이용하는 방법

```python
temps = [28,31,33,35,27,26,25]

for i in range(len((temps))):
    print(temps[i],end='.')
```

-for-each 루프를 이용하는 방법

```python
temps=[28,31,33,35,27,26,25]

for element in temps:
    print(element, end='.')
```

## 요소 삭제하기<br>
-항목이 저장된 위치를 알고 싶다면 pop(i)을 사용한다.<br>-항목의 값만 알고 있다면 remove(value)를 사용한다.

```python
fruits = ["apple","banana","grape"]
item = fruits.pop(0)
print(fruits)

fruits = ["apple","banana","grape"]
fruits.remove("banana")
print(fruits)
```
